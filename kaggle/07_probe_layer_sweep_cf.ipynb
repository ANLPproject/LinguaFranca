**Settings:** Accelerator **None (CPU)** | Internet **On**. No secrets needed.

Same LR recipe as the deployed probe (shared hop12), trained on the natural train split at **every layer 0-27**, with C = 1 (deployed) and C = 0.01 (tuned).
